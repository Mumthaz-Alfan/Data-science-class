```python
categorical_cols = [
    "company_type", "website", "website_source",
    "employees_source", "revenue_source",
    "net_income_source", "market_cap_source"
]

df[categorical_cols] = df[categorical_cols].fillna("Unknown")
```

```python

df['revenue_currency'] = df['revenue_currency'].fillna('None')
df['net_income_currency'] = df['net_income_currency'].fillna('None')
df['market_cap_currency'] = df['market_cap_currency'].fillna('None')


df['revenue_year'] = df['revenue_year'].fillna(-1).astype(int)
df['net_income_year'] = df['net_income_year'].fillna(-1).astype(int)
df['employees_year'] = df['employees_year'].fillna(-1).astype(int)
```

```python

df["revenue_imputed"] = df.groupby("company_type")["revenue"].transform(
    lambda x: x.fillna(x.median())
)

df["market_cap_missing"] = df["market_cap"].isnull().astype(int)
df["net_income_missing"] = df["net_income"].isnull().astype(int)
```

```python
df["direct_subsidiaries_count"] = df["direct_subsidiaries_count"].fillna(0)
df["max_hierarchy_depth"] = df["max_hierarchy_depth"].fillna(0)
```

```python
print(df.isnull().sum())
```

```python

df["employees"] = df.groupby("company_type")["employees"].transform(
    lambda x: x.fillna(x.median())
)

df["revenue"] = df.groupby("company_type")["revenue"].transform(
    lambda x: x.fillna(x.median())
)

df["net_income"] = df.groupby("company_type")["net_income"].transform(
    lambda x: x.fillna(x.median())
)
```

```python
print(df.isnull().sum())
```

```python

cols_to_drop = [
    'enterprise_group_id', 'display_name', 'enterprise_group_name',
    'group_root_lei', 'website', 'market_cap_date',
    'company_type', 'enterprise_group_jurisdiction',
    'enterprise_group_entity_status', 'revenue_currency',
    'net_income_currency', 'market_cap_currency',
    'website_source', 'employees_source',
    'revenue_source', 'net_income_source', 'market_cap_source'
]

X = df.drop(columns=cols_to_drop)
```

```python

print(X.dtypes.value_counts())
```

```python

y = X["revenue"]


features_to_remove = [
    "revenue",
    "revenue_imputed",
    "market_cap_missing",
    "net_income_missing"
]

X_selected = X.drop(columns=features_to_remove)
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PowerTransformer

X_train, X_test, y_train, y_test = train_test_split(
    X_selected, y, test_size=0.2, random_state=42
)


scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)


pt = PowerTransformer(method='yeo-johnson')
y_train_trans = pt.fit_transform(y_train.values.reshape(-1, 1)).flatten()
y_test_trans = pt.transform(y_test.values.reshape(-1, 1)).flatten()

print(f"Training shape: {X_train_scaled.shape}")
print(f"Testing shape:  {X_test_scaled.shape}")
```

```python
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor


models = {
    "Ridge Regression": Ridge(alpha=1.0),
    "Random Forest": RandomForestRegressor(n_estimators=100, random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42),
    "XGBoost": XGBRegressor(n_estimators=100, learning_rate=0.1, random_state=42)
}


trained_models = {}

for name, model in models.items():

    if name == "Ridge Regression":
        model.fit(X_train_scaled, y_train_trans)
    else:
        model.fit(X_train, y_train)

    trained_models[name] = model
    print(f"{name} trained successfully.")
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

results = []

for name, model in trained_models.items():

    if name == "Ridge Regression":

        y_pred_trans = model.predict(X_test_scaled)
        y_pred = pt.inverse_transform(y_pred_trans.reshape(-1, 1)).flatten()
    else:
        y_pred = model.predict(X_test)


    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    results.append({
        "Model": name,
        "RMSE": rmse,
        "MAE": mae,
        "R2 Score": r2
    })


comparison_df = pd.DataFrame(results).sort_values(by="R2 Score", ascending=False)
print("--- Model Performance Comparison ---")
print(comparison_df.to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns


best_model_name = comparison_df.iloc[0]["Model"]
best_model = trained_models[best_model_name]


if best_model_name == "Ridge Regression":
    best_pred_trans = best_model.predict(X_test_scaled)
    best_pred = pt.inverse_transform(best_pred_trans.reshape(-1, 1)).flatten()
else:
    best_pred = best_model.predict(X_test)


plt.figure(figsize=(8, 6))
sns.scatterplot(x=y_test, y=best_pred, alpha=0.7, color="indigo")


max_val = max(y_test.max(), best_pred.max())
min_val = min(y_test.min(), best_pred.min())
plt.plot([min_val, max_val], [min_val, max_val], 'r--', label="Perfect Prediction")

plt.xscale("log")
plt.yscale("log")
plt.xlabel("Actual Revenue (Log Scale)")
plt.ylabel("Predicted Revenue (Log Scale)")
plt.title(f"Actual vs. Predicted Revenue ({best_model_name})")
plt.legend()
plt.tight_layout()
plt.show()
```

```python

if hasattr(best_model, "feature_importances_"):
    importances = pd.Series(best_model.feature_importances_, index=X_selected.columns)
    top_features = importances.sort_values(ascending=False).head(10)

    plt.figure(figsize=(10, 5))
    sns.barplot(x=top_features.values, y=top_features.index, palette="viridis")
    plt.xlabel("Feature Importance Score")
    plt.ylabel("Feature")
    plt.title(f"Top 10 Most Influential Features ({best_model_name})")
    plt.tight_layout()
    plt.show()
else:

    coefs = pd.Series(np.abs(best_model.coef_), index=X_selected.columns)
    top_features = coefs.sort_values(ascending=False).head(10)

    plt.figure(figsize=(10, 5))
    sns.barplot(x=top_features.values, y=top_features.index, palette="mako")
    plt.xlabel("Absolute Coefficient Value")
    plt.ylabel("Feature")
    plt.title(f"Top 10 Most Influential Features ({best_model_name})")
    plt.tight_layout()
    plt.show()
```
